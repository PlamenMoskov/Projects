```python
import pandas as pd
import urllib.request
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import GridSearchCV
```

```python
# Download dataset
##url = "https://github.com/dataprofessor/data/raw/master/BostonHousing.csv"
filename = "BostonHousing.csv"
##urllib.request.urlretrieve(url, filename)
BostonHousing = pd.read_csv(filename)
```

```python

# Split data
Y = BostonHousing['medv']
X = BostonHousing.drop(['medv'], axis=1)
```

```python

# Train-test split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python

# Check feature correlation
plt.figure(figsize=(10, 8))
sns.heatmap(X.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Feature Correlation Matrix")
plt.show()
```

```python

# Plot feature distributions
X.hist(figsize=(12, 10), bins=20)
plt.suptitle("Feature Distributions")
plt.show()
```

```python

# Baseline model using DummyRegressor
dummy = DummyRegressor(strategy='mean')
dummy.fit(X_train, Y_train)
Y_dummy_pred = dummy.predict(X_test)
print(f"Dummy R^2 Score: {r2_score(Y_test, Y_dummy_pred):.2f}")
```

```python

# Define a pipeline with scaling and linear regression
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LinearRegression())
])


# Train and evaluate the pipeline
pipeline.fit(X_train, Y_train)
Y_pred = pipeline.predict(X_test)
print(f"Linear Regression R^2 Score: {r2_score(Y_test, Y_pred):.2f}")
```

```python
# Try different models
models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(),
    "Lasso Regression": Lasso(),
    "Random Forest": RandomForestRegressor()
}

for name, model in models.items():
    pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('model', model)
    ])
    pipe.fit(X_train, Y_train)
    Y_pred = pipe.predict(X_test)
    print(f"{name} R^2 Score: {r2_score(Y_test, Y_pred):.2f}")
```

```python
# Hyperparameter tuning for Ridge regression
param_grid = {'model__alpha': [0.1, 1, 10, 100]}
ridge_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('model', Ridge())
])

grid_search = GridSearchCV(ridge_pipe, param_grid, cv=5, scoring='r2')
grid_search.fit(X_train, Y_train)
print(f"Best Ridge Alpha: {grid_search.best_params_}")
print(f"Best Ridge R^2 Score: {grid_search.best_score_:.2f}")
```

```python




# Plot residuals
residuals = Y_test - Y_pred
sns.histplot(residuals, bins=20, kde=True)
plt.xlabel("Residuals")
plt.ylabel("Frequency")
plt.title("Residual Distribution")
plt.show()

sns.scatterplot(x=Y_test, y=Y_pred)
plt.xlabel("Actual Prices")
plt.ylabel("Predicted Prices")
plt.title("Actual vs. Predicted Prices")
plt.show()
```
